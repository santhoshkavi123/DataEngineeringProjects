# Extract Data From the Addresses

1. Query Addresses file using "read_files" function
2.  Create Addresses View in the Bronze Schema


## 1. Query Addresses file using 'read_file' function


```sql
SELECT * FROM read_files('/Volumes/gizmobox/landing/operational_data/addresses/', 
      format => 'csv',
      delimiter => '\t', 
      header => 'true'
)
```



## 2. Create Addresses View in the Bronze Schema

```sql
DROP VIEW IF EXISTS gizmobox.bronze.v_address
```

```sql
CREATE OR REPLACE VIEW gizmobox.bronze.v_address
AS
SELECT * FROM read_files('/Volumes/gizmobox/landing/operational_data/addresses/', 
  format => 'csv', 
  delimiter => '\t', 
  header => 'true'
)
```

```sql
SELECT * FROM gizmobox.bronze.v_address;
```
