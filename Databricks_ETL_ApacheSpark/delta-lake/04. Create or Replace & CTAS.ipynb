## 1. Difference between Create or Replace and Drop and Create Table statements



##### Behaviour of the Drop and Create Statements

```python
%sql
DROP TABLE IF EXISTS demo.delta_lake.companies;

CREATE TABLE demo.delta_lake.companies(
  company_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1),
  company_name STRING NOT NULL, -- NOT NULL IS A CONSTRAINT
  founded_date DATE COMMENT 'The date when the company founded', -- At column level we can write the comments
  founded_year INT GENERATED ALWAYS AS (YEAR(founded_date)) COMMENT 'The year when the company founded a computed field', 
  country STRING
)
COMMENT 'This table contains information about some of the successful tech companies'
TBLPROPERTIES('delta.appendOnly'='true');


INSERT INTO demo.delta_lake.companies (company_name, founded_date, country)
VALUES ("Apple", "1976-04-01", "USA"),
      ("Microsoft", "1975-04-04", "USA"), 
      ("Google", "1998-09-04", "USA"), 
      ("Amazon", "1994-07-05", "USA");
```

```python
%sql 


-- Even if I run multiple times above cell, the transaction logs remains same because whenever we perform DROP entire logs are deleted 
DESCRIBE HISTORY demo.delta_lake.companies
```

```python
%sql

DROP TABLE IF EXISTS demo.delta_lake.companies;
```

```python
%sql

CREATE OR REPLACE TABLE demo.delta_lake.companies(
  company_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1),
  company_name STRING NOT NULL, -- NOT NULL IS A CONSTRAINT
  founded_date DATE COMMENT 'The date when the company founded', -- At column level we can write the comments
  founded_year INT GENERATED ALWAYS AS (YEAR(founded_date)) COMMENT 'The year when the company founded a computed field', 
  country STRING
);

INSERT INTO demo.delta_lake.companies (company_name, founded_date, country)
VALUES ("Apple", "1976-04-01", "USA"),
      ("Microsoft", "1975-04-04", "USA"), 
      ("Google", "1998-09-04", "USA"), 
      ("Amazon", "1994-07-05", "USA");
```

```python
%sql

-- when I rerun the above cell multiple times, every time we have two new transaction logs we can observe
DESCRIBE HISTORY demo.delta_lake.companies
```


## 2. CTAS (Create Table As Select) Statement

```python
%sql
DROP TABLE IF EXISTS demo.delta_lake.companies_china;
```

```python
%sql

CREATE TABLE demo.delta_lake.companies_china
AS
SELECT CAST(company_id AS INT) AS company_id, 
      company_name, 
      founded_date, 
      country
FROM demo.delta_lake.companies
WHERE country = 'China';

-- You can't be doing most the operations that we are doing previously, here the options are limited but there turn-arounds in this case
```
