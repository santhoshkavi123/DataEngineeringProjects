
# History and Time Travel

1. Query Delta Lake History
2. Query Previous versions of the data 
3. Query data from a specific time
4. Restore data to a specific version.


## 1. Query Delta Lake History

```python
%sql

DESCRIBE HISTORY demo.delta_lake.companies;

-- We can see that the table has been updated twice and each update has a different version number
```


## 2. Query Data from a Specific Version

```python
%sql

SELECT * FROM demo.delta_lake.companies
VERSION AS OF 0; -- During the first operation, we have just created the table because of it we don't see any records
```

```python
%sql

SELECT * FROM demo.delta_lake.companies
VERSION AS OF 1; -- During the second operation, we have updated with one record and we can see the records
```

```python
%sql

SELECT * FROM demo.delta_lake.companies
VERSION AS OF 2; -- delta tables always provide the latest version of the table 
```


## 3. Query data for a specific time

```python
%sql

SELECT * FROM demo.delta_lake.companies
TIMESTAMP AS OF '2026-02-03T00:07:31.000+00:00';
-- We can also get the records based on the transaction log timestamp as well
```


## 4. Restore Data in the Table to a Specific Version

```python
%sql

RESTORE TABLE demo.delta_lake.companies TO VERSION AS OF 1;
```

```python
%sql

DESCRIBE HISTORY demo.delta_lake.companies;

-- We can see transaction log of restoring back to version 1 created a new version.
```

```python
%sql 

DESCRIBE EXTENDED demo.delta_lake.companies;
```

```python
%sql
SELECT * FROM demo.delta_lake.companies
```
