
# Understanding Delta Lake Trasaction Log

Understand the cloud storage directory structure behind delta lake table


##0. Create a new schema under the demo catalog for this section of the course (delta_lake)

```python
%sql
CREATE SCHEMA IF NOT EXISTS demo.delta_lake
MANAGED LOCATION 's3://databricks-catalog-sans/delta_lake'
```

```python
%sql

USE CATALOG demo; 

SHOW SCHEMAS;
```


## 1. Create a delta Lake Table

```python
%sql
CREATE OR REPLACE TABLE demo.delta_lake.companies
(
  company_name STRING, 
  founded_date DATE, 
  country STRING
);
```

```python
%sql

-- now look at extended display of the table
DESCRIBE EXTENDED demo.delta_lake.companies

-- You can post performing the CREATE a delta log is created in the managed location and let's also in the following operations how many such transaction logs are gonna create
```


## 2. Insert some data

```python
%sql

INSERT INTO demo.delta_lake.companies
VALUES ("Apples", "1976-04-01", "USA"); 
```

```python
%sql

SELECT * FROM demo.delta_lake.companies; 
```

```python
%sql 

DESCRIBE EXTENDED demo.delta_lake.companies;

-- Now, we can see one more transaction log file created in the managed location
```


## 3. Insert some more data

```python
%sql
INSERT INTO demo.delta_lake.companies
VALUES ("Microsoft", "1975-04-04", "USA"), 
        ("Google", "1998-09-04", "USA"), 
        ("Amazon", "1994-07-05", "USA");

-- Now, we can see two more transaction log files created in the managed location
-- We can also see that the table has been updated with the new records
-- Every time we perform an operation on the table, a new transaction log file is created in the managed location
```

```python
%sql
SELECT * FROM demo.delta_lake.companies;
```
