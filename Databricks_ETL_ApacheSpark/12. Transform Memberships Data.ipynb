
# Transform Memberships Data

1. Extract customer_id from the file_path
2. Write transformed data to the silver schema

```sql
SELECT * FROM gizmobox.bronze.v_memberships
```


## 1. Extract customer_id from the file_path 

```sql
-- '\\.' is required because HIVESQl treates this as '.' as regrex function
SELECT split(customer_id, '\\.')[0] as customer_id, 
        content AS membership_card
FROM (
        SELECT *, 
              split(path, '/')[7] AS customer_id
        FROM gizmobox.bronze.v_memberships
) A 
```


## 2. Write transformed data to silver schema

```sql
CREATE TABLE IF NOT EXISTS gizmobox.silver.memberships
AS
SELECT split(customer_id, '\\.')[0] as customer_id, 
        content AS membership_card
FROM (
        SELECT *, 
              split(path, '/')[7] AS customer_id
        FROM gizmobox.bronze.v_memberships
) A 
```

```sql
SELECT * FROM gizmobox.silver.memberships;
```
