
## 3. Profile Data Manually

```sql
%sql

SELECT COUNT(*), COUNT(customer_id), count(email), count(telephone) FROM gizmobox.bronze.v_customers;
```


### 3.1 COUNT_IF Function

```sql
SELECT COUNT(*), COUNT_IF(customer_id IS NULL), COUNT_IF(email IS NULL), COUNT_IF(telephone IS NULL) 
FROM gizmobox.bronze.v_customers;
```


### 3.2 WHERE function

```sql
-- Similarly we can use the where clause instead of the COUNT_IF function, depends upon the complexity of the data we can choose either to go with the count_if or the where clause


-- In the below example, we are checking if the values of the customer id is null or not, similarly we have to check if the other columns having the null values are not 
SELECT COUNT(*) FROM gizmobox.bronze.v_customers
WHERE customer_id IS NULL;

```


### Check if there any duplicates in our table

```sql
SELECT count(*) total_number_of_records, 
        COUNT(DISTINCT customer_id) unique_customer_ids
FROM gizmobox.bronze.v_customers;
```
