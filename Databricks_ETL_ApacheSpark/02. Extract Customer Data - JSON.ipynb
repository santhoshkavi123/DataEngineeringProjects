# 2. Register Files in UC - bronze schema using Views

```sql
CREATE OR REPLACE VIEW gizmobox.bronze.v_customers 
AS
SELECT *, _metadata.file_path as file_path
FROM json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers/`; 
```

```sql
SELECT * FROM gizmobox.bronze.v_customers

-- In both the cases the number of records are same
```


# Temporary Views in Databricks

They last until the current session is active, let's say if you detach the notebook from the current cluster, we can't access the temporary view even though when you re-attach to the cluster

```sql
-- One more interested thing is we can create a view on top view
CREATE OR REPLACE TEMPORARY VIEW tv_customers
AS
SELECT * FROM gizmobox.bronze.v_customers;
```

```sql
SELECT * FROM tv_customers;
```


# Global Temporary View in Databricks

In this case, even the notebook is detach from the current cluster, we can access the global temporary view tables when you re-attach.

```sql
CREATE OR REPLACE VIEW gtv_customers
AS
SELECT * FROM gizmobox.bronze.v_customers;
-- Gotcha this feature is not supported in serverless compute but it works 
```
