
# Set-up the project environment for GizmoBox Data Lakehouse




## 1. Access the s3 bucket gizmobox

```sql
%fs ls 's3://databricks-catalog-sans/gizmobox/landing/'
```


## 2. Create the catalog - gizmobox

```sql
-- first, let's see what are the catalogs are present 

SHOW CATALOGS;
```

```sql
-- make sure we mention the managed location so, that databricks stores the information in external location - S3 bucket
CREATE CATALOG IF NOT EXISTS gizmobox
MANAGED LOCATION 's3://databricks-catalog-sans/'
COMMENT 'This is the catalog for the Gizmobox Data lakehouse';
```

```sql
-- Let's check if the catalog is created!!

SHOW CATALOGS;
```


## 3. Create Schemas

1. Landing
2. Bronze
3. Silver
4. Gold

```sql
-- You can check what is the current catalog 

SELECT current_catalog();
```


### 1. Landing Schema for Gizmobox Data Lakehouse

```sql
-- Now, change catalog to gizmobox

USE CATALOG gizmobox;

SELECT current_catalog();

---Now, let's create landing schema
CREATE SCHEMA IF NOT EXISTS landing
MANAGED LOCATION 's3://databricks-catalog-sans/landing'
COMMENT 'This is the landing schema for the Gizmobox Data lakehouse';
```


### 2. Bronze Schema for the Gizmobox Data Lakehouse

```sql
CREATE SCHEMA IF NOT EXISTS bronze
MANAGED LOCATION 's3://databricks-catalog-sans/bronze'
COMMENT 'This is the bronze schema for the Gizmobox Data lakehouse';
```

### 3. Silver Schema for the Gizmobox Data Lakehouse

```sql
CREATE SCHEMA IF NOT EXISTS silver
MANAGED LOCATION 's3://databricks-catalog-sans/silver'
COMMENT 'This is the silver schema for the Gizmobox Data Lakehouse';
```

### 4. Gold Schema for the Gizmobox Data Lakehouse

```sql
CREATE SCHEMA IF NOT EXISTS gold
MANAGED LOCATION 's3://databricks-catalog-sans/gold'
COMMENT 'This is the gold schema for the Gizmobox Data Lakehouse';
```

```sql
-- Show the schemas that are present in the catalog
SHOW SCHEMAS
```

## 4. Create Volume

```sql
-- we want the volumne pointing to the operation data

USE CATALOG gizmobox;

USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
LOCATION 's3://databricks-catalog-sans/gizmobox/landing/operational_data/'
COMMENT 'This is the external volume for the operational data in the landing schema';
```
