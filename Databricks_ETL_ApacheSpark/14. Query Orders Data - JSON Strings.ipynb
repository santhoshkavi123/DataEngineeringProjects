
# Query Orders Data as JSON Strings

1. Extract Top level Column Values
2. Extract Array element
3. Extract Nested Column Values
4. CAST Column Values to Specific Data Type

```sql
SELECT * FROM gizmobox.bronze.v_orders;
```


## 1. Extract Top level Column Values

    Top Level Object <column_name>:<extraction_path>

```sql
SELECT value:order_id as order_id, value
FROM gizmobox.bronze.v_orders
```


## 2. Extract Array element
    To extract array element - use the indices in brackets <column_name>:<extraction_path>[index]

```sql
SELECT value:items[0] as item_id_1, 
       value:items[1] as item_id_2, 
       value
FROM gizmobox.bronze.v_orders
```


## 3. Extract Nested Column Values
    To extract nested columns, use dot notation <column_name>:<extraction_path>[index].<leaf_node>

```sql
SELECT value:items[0] as item_id_1_arr, 
       value:items[1] as item_id_2_arr,
       value:items[0].item_id as item_id_1_Acc_nested_col, 
       value:items[1].item_id as item_id_1_Acc_nested_col,
       value
FROM gizmobox.bronze.v_orders
```


## 4. CAST Column Values to a Specific Data Type

We can understand that Databricks provides the facility of handling of JSON strings, but we have to deal with a lot of complexity.

```sql
SELECT value:items[0] as item_id_1_arr, 
       value:items[1] as item_id_2_arr,
       value:items[0].item_id::INTEGER as item_id_1_Acc_nested_col, 
       value:items[1].item_id::INTEGER as item_id_1_Acc_nested_col,
       value
FROM gizmobox.bronze.v_orders
```
