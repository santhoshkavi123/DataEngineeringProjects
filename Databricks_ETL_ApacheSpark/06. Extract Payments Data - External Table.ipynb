
## 2. Create External Table

```sql
CREATE TABLE IF NOT EXISTS gizmobox.bronze.ext_payments(
  payment_id INTEGER, 
  order_id INTEGER, 
  payment_timestamp TIMESTAMP, 
  payment_status INTEGER, 
  payment_method STRING
)
USING CSV
OPTIONS (
  header = 'true', 
  delimiter = ','
)
LOCATION 's3://databricks-catalog-sans/gizmobox/landing/external_data/payments/' -- when location is mentioned it means we are creating external table
```

```sql
SELECT * FROM gizmobox.bronze.ext_payments;
```

```sql
DESCRIBE EXTENDED gizmobox.bronze.ext_payments;
```


## 3. What to do if we some files at the cloud storage is deleted, we need to refresh the data 

```sql
REFRESH TABLE gizmobox.bronze.ext_payments;

-- This basically updates the metadata of the table, if we don't run this command before running the select statement 
-- 1. Refresh the external table, 2. Now perform an SELECT statement 
```
